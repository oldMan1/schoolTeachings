```python
from collections.abc import Iterable, Mapping
from typing import Any

def print_symbols(rows: Iterable[Mapping[str, Any]]) -> None:
    # your code here
    for row in rows:
        symbol = row.get('symbol')
        if symbol is None:
            print("missing symbol")
        else:
            print(symbol)

print_symbols(rows)
    
```
