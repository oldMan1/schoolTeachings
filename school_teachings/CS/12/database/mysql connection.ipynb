# 1. Integrate SQL with Python by importing a suitable module and connect to MySQL.

Change the `password` value if your MySQL root password is different.


```python
import mysql.connector  # Import mysql.connector to connect Python with MySQL.

host_name = "127.0.0.1"  # Store the MySQL server address.
user_name = "root"  # Store the MySQL user name.
password_value = ""  # Store the MySQL password; change it if required.
database_name = "class12_practical"  # Store the database name for this practical.

connection = mysql.connector.connect(  # Create a connection between Python and MySQL.
    host=host_name,  # Pass the host name to the connection.
    user=user_name,  # Pass the user name to the connection.
    password=password_value  # Pass the password to the connection.
)  # End the connection statement.

cursor = connection.cursor()  # Create a cursor object to run SQL commands.
cursor.execute("CREATE DATABASE IF NOT EXISTS " + database_name)  # Create the database if it does not exist.
cursor.execute("USE " + database_name)  # Select the database for the practical work.
print("Connected to MySQL and using database:", database_name)  # Display a success message.
```

# 2. Create a student table and insert data.


```python
def show_records(sql_query):  # Define a function to display records for a SELECT query.
    cursor.execute(sql_query)  # Execute the SELECT query received by the function.
    records = cursor.fetchall()  # Fetch all rows returned by the query.
    for record in records:  # Read each row one by one.
        print(record)  # Display the current row.

cursor.execute("DROP TABLE IF EXISTS student")  # Remove old student table so the notebook can run again cleanly.

create_table_query = """
CREATE TABLE student (
    roll_no INT PRIMARY KEY,
    name VARCHAR(30),
    class_section VARCHAR(5),
    city VARCHAR(30),
    marks INT
)
"""  # Store the CREATE TABLE command in a variable.

cursor.execute(create_table_query)  # Create the student table.

insert_query = "INSERT INTO student VALUES (%s, %s, %s, %s, %s)"  # Store the INSERT command with placeholders.

student_records = [  # Create a list of student records.
    (1, "Aman", "12A", "Delhi", 88),  # Store first student record.
    (2, "Riya", "12A", "Mumbai", 76),  # Store second student record.
    (3, "Kabir", "12B", "Delhi", 64),  # Store third student record.
    (4, "Zoya", "12B", "Kolkata", 69),  # Store fourth student record.
    (5, "Isha", "12A", "Mumbai", 92),  # Store fifth student record.
    (6, "Neel", "12B", "Delhi", 81)  # Store sixth student record.
]  # End the list of student records.

cursor.executemany(insert_query, student_records)  # Insert all student records into the table.
connection.commit()  # Save the inserted records permanently in the database.

print("Student table created and records inserted.")  # Display a success message.
show_records("SELECT * FROM student")  # Display all records from the student table.
```

# 3. Use ALTER TABLE to add a new attribute, modify a data type, and drop an attribute.


```python
cursor.execute("ALTER TABLE student ADD email VARCHAR(50)")  # Add a new column named email.
cursor.execute("ALTER TABLE student MODIFY marks FLOAT")  # Change the data type of marks from INT to FLOAT.
cursor.execute("ALTER TABLE student DROP COLUMN email")  # Remove the email column from the table.
connection.commit()  # Save the table structure changes.

print("Table structure after ALTER commands:")  # Display a heading for table structure.
cursor.execute("DESC student")  # Run DESC command to show the structure of the table.
structure = cursor.fetchall()  # Fetch the table structure rows.
for column in structure:  # Read each column detail one by one.
    print(column)  # Display the current column detail.
```

# 4. Use UPDATE to modify data in the student table.


```python
cursor.execute("UPDATE student SET marks = 95 WHERE roll_no = 3")  # Update marks for the student whose roll number is 3.
connection.commit()  # Save the updated marks in the database.

print("Records after updating marks of roll number 3:")  # Display a heading for updated records.
show_records("SELECT * FROM student")  # Display all records after the update.
```

# 5. Use ORDER BY to display data in ascending and descending order.


```python
print("Records in ascending order of marks:")  # Display a heading for ascending order.
show_records("SELECT * FROM student ORDER BY marks ASC")  # Display records from lowest marks to highest marks.

print("Records in descending order of marks:")  # Display a heading for descending order.
show_records("SELECT * FROM student ORDER BY marks DESC")  # Display records from highest marks to lowest marks.
```

# 6. Use DELETE to remove tuple(s) from the student table.


```python
cursor.execute("DELETE FROM student WHERE marks < 70")  # Delete students whose marks are less than 70.
connection.commit()  # Save the delete operation in the database.

print("Records after deleting students with marks less than 70:")  # Display a heading for remaining records.
show_records("SELECT * FROM student")  # Display all remaining records.
```

# 7. Use GROUP BY to find MIN, MAX, SUM, COUNT, and AVG.


```python
group_query = """
SELECT city,
       MIN(marks),
       MAX(marks),
       SUM(marks),
       COUNT(*),
       AVG(marks)
FROM student
GROUP BY city
"""  # Store the GROUP BY query in a variable.

cursor.execute(group_query)  # Execute the GROUP BY query.
result = cursor.fetchall()  # Fetch all rows from the result.

print("City wise MIN, MAX, SUM, COUNT, and AVG of marks:")  # Display a heading for grouped result.
for row in result:  # Read each grouped row one by one.
    print(row)  # Display the current grouped row.
```

# 8. Close the MySQL cursor and connection.


```python
cursor.close()  # Close the cursor object.
connection.close()  # Close the MySQL connection.
print("MySQL connection closed.")  # Display a closing message.
```
