# 23. Write a program using a user defined function `myMean()` to calculate the mean of floating values stored in a list.


```python
def myMean(values):  # Define a function named myMean that accepts a list of values.
    total = 0.0  # Store the running total as a floating point number.
    count = 0  # Store how many values are present in the list.
    for value in values:  # Visit each value one by one from the list.
        total = total + value  # Add the current value to the running total.
        count = count + 1  # Increase the count by one for the current value.
    if count == 0:  # Check whether the list is empty.
        return 0  # Return 0 because mean cannot be calculated for an empty list.
    mean = total / count  # Divide the total by the count to calculate mean.
    return mean  # Send the calculated mean back to the calling statement.

mylist = [1.0, 2.0, 3.0, 4.0, 5.5]  # Store floating point values in a list.
result = myMean(mylist)  # Call the function and store the returned mean.
print("Mean is:", result)  # Display the final mean.
```

# 24. Write a program using a user defined function `calcFact()` to calculate and display the factorial of a number `num` passed as an argument.


```python
def calcFact(num):  # Define a function named calcFact that accepts one number.
    factorial = 1  # Store 1 as the starting value of factorial.
    for value in range(1, num + 1):  # Run the loop from 1 up to the given number.
        factorial = factorial * value  # Multiply the current value with the factorial.
    return factorial  # Return the calculated factorial.

num = int(input("Enter a number: "))  # Accept the number from the user.
if num < 0:  # Check whether the number is negative.
    print("Factorial is not defined for negative numbers.")  # Display an error message for negative input.
else:  # Run this block when the number is zero or positive.
    answer = calcFact(num)  # Call the function and store the factorial.
    print("Factorial of", num, "is", answer)  # Display the final answer.
```

# 29. Write a program that implements a user defined function that accepts Principal Amount, Rate, Time, and Number of Times the interest is compounded to calculate and display compound interest.


```python
def calculateCompoundInterest(principal, rate, time, times_compounded):  # Define a function for compound interest.
    rate_decimal = rate / 100  # Convert rate percentage into decimal form.
    amount = principal * (1 + rate_decimal / times_compounded) ** (times_compounded * time)  # Calculate final amount.
    compound_interest = amount - principal  # Subtract principal from amount to get compound interest.
    return compound_interest  # Return the compound interest.

principal = float(input("Enter principal amount: "))  # Accept principal amount from the user.
rate = float(input("Enter rate of interest: "))  # Accept annual rate of interest from the user.
time = float(input("Enter time in years: "))  # Accept time in years from the user.
times_compounded = int(input("Enter number of times interest is compounded per year: "))  # Accept compounding frequency.
ci = calculateCompoundInterest(principal, rate, time, times_compounded)  # Call the function and store compound interest.
print("Compound Interest is:", ci)  # Display the calculated compound interest.
```

# 30. Write a Python program to convert a real decimal number with both integer and fractional parts into its binary equivalent.


```python
def integerToBinary(integer_part):  # Define a function to convert the integer part into binary.
    if integer_part == 0:  # Check whether the integer part is zero.
        return "0"  # Return 0 as the binary form of zero.
    binary_integer = ""  # Create an empty string to store binary digits.
    while integer_part > 0:  # Repeat the loop until the integer part becomes zero.
        remainder = integer_part % 2  # Find the remainder after division by 2.
        binary_integer = str(remainder) + binary_integer  # Add the remainder at the beginning of the binary string.
        integer_part = integer_part // 2  # Divide the integer part by 2 using integer division.
    return binary_integer  # Return the binary form of the integer part.

def fractionToBinary(fractional_part):  # Define a function to convert the fractional part into binary.
    binary_fraction = ""  # Create an empty string to store fractional binary digits.
    count = 0  # Store how many fractional binary digits have been generated.
    while fractional_part > 0 and count < 10:  # Repeat until fraction becomes zero or 10 digits are generated.
        fractional_part = fractional_part * 2  # Multiply the fractional part by 2.
        bit = int(fractional_part)  # Take the integer part as the next binary bit.
        binary_fraction = binary_fraction + str(bit)  # Add the bit to the binary fraction string.
        fractional_part = fractional_part - bit  # Remove the integer part from the fractional value.
        count = count + 1  # Increase the digit counter by one.
    return binary_fraction  # Return the binary form of the fractional part.

decimal_number = float(input("Enter a decimal number: "))  # Accept a decimal number from the user.
is_negative = decimal_number < 0  # Check whether the number is negative.
decimal_number = abs(decimal_number)  # Convert the number to positive for calculation.
integer_part = int(decimal_number)  # Separate the integer part of the number.
fractional_part = decimal_number - integer_part  # Separate the fractional part of the number.
binary_integer = integerToBinary(integer_part)  # Convert the integer part to binary.
binary_fraction = fractionToBinary(fractional_part)  # Convert the fractional part to binary.
if binary_fraction == "":  # Check whether there is no fractional binary part.
    binary_number = binary_integer  # Use only the integer binary part.
else:  # Run this block when a fractional binary part exists.
    binary_number = binary_integer + "." + binary_fraction  # Join integer and fractional binary parts.
if is_negative == True:  # Check whether the original number was negative.
    binary_number = "-" + binary_number  # Add the negative sign to the binary result.
print("Binary representation is:", binary_number)  # Display the final binary result.
```

# 31. Write a program that contains user defined functions to calculate area, perimeter, or surface area for square, rectangle, triangle, circle, and cylinder. The functions should accept values as parameters and return the calculated value. Import the required module and use the appropriate functions.


```python
import math  # Import the math module to use the value of pi.

def squareArea(side):  # Define a function to calculate area of a square.
    return side * side  # Return side multiplied by side.

def squarePerimeter(side):  # Define a function to calculate perimeter of a square.
    return 4 * side  # Return four times the side.

def rectangleArea(length, breadth):  # Define a function to calculate area of a rectangle.
    return length * breadth  # Return length multiplied by breadth.

def rectanglePerimeter(length, breadth):  # Define a function to calculate perimeter of a rectangle.
    return 2 * (length + breadth)  # Return two times the sum of length and breadth.

def triangleArea(base, height):  # Define a function to calculate area of a triangle.
    return 0.5 * base * height  # Return half multiplied by base and height.

def trianglePerimeter(side1, side2, side3):  # Define a function to calculate perimeter of a triangle.
    return side1 + side2 + side3  # Return the sum of all three sides.

def circleArea(radius):  # Define a function to calculate area of a circle.
    return math.pi * radius * radius  # Return pi multiplied by radius squared.

def circleCircumference(radius):  # Define a function to calculate circumference of a circle.
    return 2 * math.pi * radius  # Return two multiplied by pi and radius.

def cylinderCSA(radius, height):  # Define a function to calculate curved surface area of a cylinder.
    return 2 * math.pi * radius * height  # Return 2*pi*r*h.

def cylinderTSA(radius, height):  # Define a function to calculate total surface area of a cylinder.
    return 2 * math.pi * radius * (radius + height)  # Return 2*pi*r*(r+h).

print("1. Square")  # Display square option.
print("2. Rectangle")  # Display rectangle option.
print("3. Triangle")  # Display triangle option.
print("4. Circle")  # Display circle option.
print("5. Cylinder")  # Display cylinder option.
choice = int(input("Enter your choice: "))  # Accept the user's menu choice.

if choice == 1:  # Check whether the user selected square.
    side = float(input("Enter side: "))  # Accept side of the square.
    print("Area of square:", squareArea(side))  # Display area of the square.
    print("Perimeter of square:", squarePerimeter(side))  # Display perimeter of the square.
elif choice == 2:  # Check whether the user selected rectangle.
    length = float(input("Enter length: "))  # Accept length of the rectangle.
    breadth = float(input("Enter breadth: "))  # Accept breadth of the rectangle.
    print("Area of rectangle:", rectangleArea(length, breadth))  # Display area of the rectangle.
    print("Perimeter of rectangle:", rectanglePerimeter(length, breadth))  # Display perimeter of the rectangle.
elif choice == 3:  # Check whether the user selected triangle.
    base = float(input("Enter base: "))  # Accept base of the triangle.
    height = float(input("Enter height: "))  # Accept height of the triangle.
    side1 = float(input("Enter side 1: "))  # Accept first side of the triangle.
    side2 = float(input("Enter side 2: "))  # Accept second side of the triangle.
    side3 = float(input("Enter side 3: "))  # Accept third side of the triangle.
    print("Area of triangle:", triangleArea(base, height))  # Display area of the triangle.
    print("Perimeter of triangle:", trianglePerimeter(side1, side2, side3))  # Display perimeter of the triangle.
elif choice == 4:  # Check whether the user selected circle.
    radius = float(input("Enter radius: "))  # Accept radius of the circle.
    print("Area of circle:", circleArea(radius))  # Display area of the circle.
    print("Circumference of circle:", circleCircumference(radius))  # Display circumference of the circle.
elif choice == 5:  # Check whether the user selected cylinder.
    radius = float(input("Enter radius: "))  # Accept radius of the cylinder.
    height = float(input("Enter height: "))  # Accept height of the cylinder.
    print("Curved surface area of cylinder:", cylinderCSA(radius, height))  # Display curved surface area.
    print("Total surface area of cylinder:", cylinderTSA(radius, height))  # Display total surface area.
else:  # Run this block for an invalid menu choice.
    print("Invalid choice")  # Display an invalid choice message.
```

# 32. Write a program that creates a GK quiz consisting of any five questions. Display the questions randomly. Create a user defined function `score()` to calculate the score and another function `remark(scorevalue)` to display remarks.


```python
import random  # Import the random module to shuffle the quiz questions.

questions = [  # Create a list to store all quiz questions and answers.
    {"question": "What is the capital of India?", "answer": "new delhi"},  # Store question 1 with answer.
    {"question": "How many states are there in India?", "answer": "28"},  # Store question 2 with answer.
    {"question": "Which planet is known as the Red Planet?", "answer": "mars"},  # Store question 3 with answer.
    {"question": "Who wrote the national anthem of India?", "answer": "rabindranath tagore"},  # Store question 4 with answer.
    {"question": "Which gas do plants absorb during photosynthesis?", "answer": "carbon dioxide"}  # Store question 5 with answer.
]  # Close the list of questions.

def score(questions):  # Define a function to calculate quiz score.
    marks = 0  # Store the initial score as zero.
    random.shuffle(questions)  # Shuffle the questions so they appear randomly.
    for item in questions:  # Read each question dictionary one by one.
        print(item["question"])  # Display the question.
        user_answer = input("Enter answer: ")  # Accept the answer from the user.
        user_answer = user_answer.lower()  # Convert the answer to lowercase for comparison.
        if user_answer == item["answer"]:  # Compare the user's answer with the correct answer.
            marks = marks + 1  # Increase score by one for a correct answer.
    return marks  # Return the final score.

def remark(scorevalue):  # Define a function to display remarks based on score.
    if scorevalue == 5:  # Check whether the score is 5.
        print("Outstanding")  # Display remark for score 5.
    elif scorevalue == 4:  # Check whether the score is 4.
        print("Excellent")  # Display remark for score 4.
    elif scorevalue == 3:  # Check whether the score is 3.
        print("Good")  # Display remark for score 3.
    elif scorevalue == 2:  # Check whether the score is 2.
        print("Read more to score more")  # Display remark for score 2.
    elif scorevalue == 1:  # Check whether the score is 1.
        print("Needs to take interest")  # Display remark for score 1.
    else:  # Run this block when score is 0.
        print("General knowledge will always help you. Take it seriously.")  # Display remark for score 0.

final_score = score(questions)  # Call the score function and store the final score.
print("Your score is:", final_score)  # Display the final score.
remark(final_score)  # Call the remark function to display the final remark.
```

# 34. Read a text file and display the number of vowels, consonants, uppercase characters, and lowercase characters in the file.


```python
file_name = "q1.txt"  # Store the name of the text file.
sample_text = "Python Is Easy\nComputer Science CLASS 12"  # Store sample text for counting characters.

with open(file_name, "w") as file_object:  # Open the file in write mode to create sample data.
    file_object.write(sample_text)  # Write the sample text into the file.

vowels = "aeiouAEIOU"  # Store all vowel characters for checking.
vowel_count = 0  # Store the number of vowels.
consonant_count = 0  # Store the number of consonants.
uppercase_count = 0  # Store the number of uppercase characters.
lowercase_count = 0  # Store the number of lowercase characters.

with open(file_name, "r") as file_object:  # Open the file in read mode.
    for line in file_object:  # Read each line from the file.
        for char in line:  # Read each character from the current line.
            if char.isalpha():  # Check whether the character is an alphabet.
                if char in vowels:  # Check whether the alphabet is a vowel.
                    vowel_count = vowel_count + 1  # Increase the vowel count by one.
                else:  # Run this block when the alphabet is not a vowel.
                    consonant_count = consonant_count + 1  # Increase the consonant count by one.
                if char.isupper():  # Check whether the alphabet is uppercase.
                    uppercase_count = uppercase_count + 1  # Increase the uppercase count by one.
                if char.islower():  # Check whether the alphabet is lowercase.
                    lowercase_count = lowercase_count + 1  # Increase the lowercase count by one.

print("Vowels:", vowel_count)  # Display the total number of vowels.
print("Consonants:", consonant_count)  # Display the total number of consonants.
print("Uppercase characters:", uppercase_count)  # Display the total number of uppercase characters.
print("Lowercase characters:", lowercase_count)  # Display the total number of lowercase characters.
```

# 36. Create a binary file with name and roll number. Search for a given roll number and display the name. If it is not found, display an appropriate message.


```python
import pickle  # Import pickle module to work with binary files.

file_name = "student_roll.dat"  # Store the name of the binary file.
students = [  # Create a list of student records.
    {"roll_no": 1, "name": "Aman"},  # Store first student record.
    {"roll_no": 2, "name": "Riya"},  # Store second student record.
    {"roll_no": 3, "name": "Kabir"}  # Store third student record.
]  # Close the list of records.

with open(file_name, "wb") as file_object:  # Open the binary file in write binary mode.
    pickle.dump(students, file_object)  # Write the student list into the binary file.

roll_to_search = int(input("Enter roll number to search: "))  # Accept the roll number to search.
found = False  # Store False because the record is not found yet.

with open(file_name, "rb") as file_object:  # Open the binary file in read binary mode.
    stored_students = pickle.load(file_object)  # Read the student list from the binary file.

for student in stored_students:  # Read each student record one by one.
    if student["roll_no"] == roll_to_search:  # Check whether roll number matches the search value.
        print("Name is:", student["name"])  # Display the name of the matching student.
        found = True  # Change found to True because the record is found.

if found == False:  # Check whether no matching record was found.
    print("Roll number not found")  # Display the not found message.
```

# 37. Create a binary file with roll number, name, and marks. Input a roll number and update the marks.


```python
import pickle  # Import pickle module to work with binary files.

file_name = "student_marks.dat"  # Store the name of the binary file.
students = [  # Create a list of student records.
    {"roll_no": 1, "name": "Aman", "marks": 85},  # Store first student record.
    {"roll_no": 2, "name": "Riya", "marks": 91},  # Store second student record.
    {"roll_no": 3, "name": "Kabir", "marks": 78}  # Store third student record.
]  # Close the list of records.

with open(file_name, "wb") as file_object:  # Open the binary file in write binary mode.
    pickle.dump(students, file_object)  # Write the student records into the binary file.

roll_to_update = int(input("Enter roll number to update marks: "))  # Accept the roll number to update.
new_marks = int(input("Enter new marks: "))  # Accept the new marks.
found = False  # Store False because the record is not updated yet.

with open(file_name, "rb") as file_object:  # Open the binary file in read binary mode.
    stored_students = pickle.load(file_object)  # Read the student records from the file.

for student in stored_students:  # Read each student record one by one.
    if student["roll_no"] == roll_to_update:  # Check whether roll number matches the input roll number.
        student["marks"] = new_marks  # Update marks for the matching student.
        found = True  # Change found to True because the record was updated.

with open(file_name, "wb") as file_object:  # Open the binary file again in write binary mode.
    pickle.dump(stored_students, file_object)  # Write the updated student records back to the file.

if found == True:  # Check whether the update was successful.
    print("Marks updated successfully")  # Display success message.
else:  # Run this block when roll number was not found.
    print("Roll number not found")  # Display not found message.

print("Updated records are:")  # Display a heading for updated records.
for student in stored_students:  # Read each updated student record.
    print(student)  # Display the current student record.
```

# 39. Write a Python program to implement a stack using list.


```python
stack = []  # Create an empty list to store stack elements.

while True:  # Start an infinite loop for the menu.
    print("1. Push")  # Display push option.
    print("2. Pop")  # Display pop option.
    print("3. Peek")  # Display peek option.
    print("4. Display")  # Display stack option.
    print("5. Exit")  # Display exit option.
    choice = int(input("Enter your choice: "))  # Accept the user's menu choice.

    if choice == 1:  # Check whether the user selected push.
        value = input("Enter value to push: ")  # Accept the value to insert.
        stack.append(value)  # Add the value at the top of the stack.
        print("Value pushed")  # Display push success message.
    elif choice == 2:  # Check whether the user selected pop.
        if len(stack) == 0:  # Check whether the stack is empty.
            print("Stack underflow")  # Display underflow message.
        else:  # Run this block when stack has elements.
            value = stack.pop()  # Remove and return the top element.
            print("Popped value:", value)  # Display the popped value.
    elif choice == 3:  # Check whether the user selected peek.
        if len(stack) == 0:  # Check whether the stack is empty.
            print("Stack is empty")  # Display empty stack message.
        else:  # Run this block when stack has elements.
            print("Top value:", stack[-1])  # Display the top element without removing it.
    elif choice == 4:  # Check whether the user selected display.
        print("Stack:", stack)  # Display the current stack.
    elif choice == 5:  # Check whether the user selected exit.
        break  # Stop the loop and end the program.
    else:  # Run this block for an invalid menu choice.
        print("Invalid choice")  # Display invalid choice message.
```

# 40. Create a CSV file by entering user-id and password. Read and search the password for a given user-id.


```python
import csv  # Import the csv module to work with CSV files.

file_name = "users.csv"  # Store the name of the CSV file.
number_of_users = int(input("Enter number of users: "))  # Accept how many user records will be entered.

with open(file_name, "w", newline="") as file_object:  # Open the CSV file in write mode.
    writer = csv.writer(file_object)  # Create a CSV writer object.
    writer.writerow(["userid", "password"])  # Write the header row in the CSV file.
    for count in range(number_of_users):  # Repeat the loop for each user record.
        userid = input("Enter user-id: ")  # Accept user-id from the user.
        password = input("Enter password: ")  # Accept password from the user.
        writer.writerow([userid, password])  # Write the user-id and password into the CSV file.

search_userid = input("Enter user-id to search password: ")  # Accept the user-id to search.
found = False  # Store False because the user-id is not found yet.

with open(file_name, "r") as file_object:  # Open the CSV file in read mode.
    reader = csv.reader(file_object)  # Create a CSV reader object.
    next(reader)  # Skip the header row.
    for row in reader:  # Read each remaining row from the CSV file.
        if row[0] == search_userid:  # Check whether the first column matches the searched user-id.
            print("Password is:", row[1])  # Display the password from the second column.
            found = True  # Change found to True because the user-id was found.

if found == False:  # Check whether the user-id was not found.
    print("User-id not found")  # Display the not found message.
```

```python
file_name = "seek_demo.txt"  # Store the name of the text file.
sample_text = "Learning file handling in Python"  # Store sample text for the file.

with open(file_name, "w") as file_object:  # Open the file in write mode.
    file_object.write(sample_text)  # Write sample text into the file.

with open(file_name, "r") as file_object:  # Open the file in read mode.
    print("Initial position:", file_object.tell())  # Display the starting position of the file pointer.
    data = file_object.read(8)  # Read the first 8 characters from the file.
    print("Data read:", data)  # Display the data that was read.
    print("Position after reading:", file_object.tell())  # Display the file pointer position after reading.
    file_object.seek(0)  # Move the file pointer back to the beginning of the file.
    print("Position after seek(0):", file_object.tell())  # Display the new file pointer position.
    file_object.seek(10)  # Move the file pointer to the 10th byte position.
    print("Position after seek(10):", file_object.tell())  # Display the file pointer position after seek.
    remaining_data = file_object.read()  # Read the remaining data from the current position.
    print("Remaining data:", remaining_data)  # Display the remaining data.
```

# 42. Write a Python function so that we can go backwards from the current file position.


```python
def moveBackward(file_object, steps):  # Define a function to move the file pointer backwards.
    current_position = file_object.tell()  # Store the current file pointer position.
    new_position = current_position - steps  # Calculate the new position by subtracting steps.
    if new_position < 0:  # Check whether the new position goes before the beginning of file.
        new_position = 0  # Set the new position to zero to stay inside the file.
    file_object.seek(new_position)  # Move the file pointer to the new position.
    return file_object.tell()  # Return the updated file pointer position.

file_name = "backward_seek.txt"  # Store the name of the text file.
sample_text = "Computer Science Practical"  # Store sample text for the file.

with open(file_name, "w") as file_object:  # Open the file in write mode.
    file_object.write(sample_text)  # Write sample text into the file.

with open(file_name, "r") as file_object:  # Open the file in read mode.
    data = file_object.read(16)  # Read the first 16 characters from the file.
    print("Data read:", data)  # Display the data that was read.
    print("Current position:", file_object.tell())  # Display the current file pointer position.
    position = moveBackward(file_object, 6)  # Move the file pointer 6 positions backwards.
    print("Position after moving backward:", position)  # Display the new file pointer position.
    print("Data from new position:", file_object.read())  # Display data from the new position.
```

# 43. Write a menu driven Python program that uses the `write()` method to store user inputs in a text file.


```python
file_name = "practice.txt"  # Store the name of the text file.

with open(file_name, "w") as file_object:  # Open the file in write mode.
    while True:  # Start an infinite loop to accept multiple lines.
        data = input("Enter data to save in the text file: ")  # Accept data from the user.
        file_object.write(data + "\n")  # Write the data followed by a newline into the file.
        answer = input("Do you wish to enter more data? (y/n): ")  # Ask whether the user wants to continue.
        answer = answer.lower()  # Convert the answer to lowercase.
        if answer == "n":  # Check whether the user wants to stop.
            break  # Stop the loop.

print("Data saved in", file_name)  # Display a confirmation message.
```
