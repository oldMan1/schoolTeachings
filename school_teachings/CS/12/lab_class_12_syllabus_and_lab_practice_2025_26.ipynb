# 24.Write a program using a user defined function calcFact() to calculate and display the factorial of a number num passed as an argument.

```python
def calcfact(factorialOf):
  if factorialOf < 0:
    print("please enter a positive number to calculate factorial")
    return
  factorial = 1
  for num in range(1, factorialOf + 1):
    factorial = factorial * num
  return factorial

print("enter a number to calculate factorial")
num = int(input())
print("user input is", num)

print("factorial of", num, "is", calcfact(num))
```

# 27. Write a program that has a user defined function to accept the coefficients of a quadratic equation in variables and calculates its determinant. For example : if the coefficients are stored in the variables a,b,c then calculate determinant as b2-4ac. Write the appropriate condition to check determinants on positive, zero and negative and output appropriate result.

```python
def calculateDeterminant():
  input1 = input("enter coefficient of determinants a,b,c seperated by space")
  inputs = input1.strip().split()
  a, b, c = inputs
  a,b,c = int(a), int(b), int(c)
  determinant = b**2 - 4*a*c
  return determinant

determinant = calculateDeterminant()

if determinant > 0:
  print("the determinant is positive. equation has two real and distinct roots.")
elif determinant < 0:
  print("the determinant is negative. equation has two complex roots.")
else:  
  print("the determinant is zero. equation has one real root.")
```

# 29. Write a program that implements a user defined function that accepts Principal Amount, Rate, Time, Number of Times the interest is compounded to calculate and displays compound interest. (Hint: CI=P*(1+r/n)***nt)

```python
def calculate_compound_interest(principal, rate, time, times_compounded):
    rate = rate / 100
    amount = principal * (1 + rate / times_compounded) ** (times_compounded * time)
    compound_interest = amount - principal
    return compound_interest


principal = float(input("Enter principal amount: "))
rate = float(input("Enter rate of interest: "))
time = float(input("Enter time in years: "))
times_compounded = int(input("Enter number of times interest is compounded per year: "))

compound_interest = calculate_compound_interest(principal, rate, time, times_compounded)

print("Compound Interest is:", compound_interest)
```

# 30. Write a Python program to convert a real (decimal) number with both integer and fractional parts into its binary equivalent.

Your program should:

* Accept a real number from the user (e.g., 10.25).

Example Input:
* Enter a decimal number: 10.25

Example Output
* Binary representation of 10.25 is: 1010.01

```python
def integer_to_binary(integer_part):
    if integer_part == 0:
        return "0"

    binary_integer = ""

    while integer_part > 0:
        remainder = integer_part % 2
        binary_integer = str(remainder) + binary_integer
        integer_part = integer_part // 2

    return binary_integer


def fraction_to_binary(fractional_part):
    binary_fraction = ""

    while fractional_part > 0:
        fractional_part = fractional_part * 2
        bit = int(fractional_part)
        binary_fraction = binary_fraction + str(bit)
        fractional_part = fractional_part - bit

    return binary_fraction


decimal_number = float(input("Enter a decimal number: "))

integer_part = int(decimal_number)
fractional_part = decimal_number - integer_part

binary_integer = integer_to_binary(integer_part)
binary_fraction = fraction_to_binary(fractional_part)

binary_number = binary_integer + "." + binary_fraction

print("Binary representation of", decimal_number, "is:", binary_number)
```
