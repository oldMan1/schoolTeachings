# File Handling in Python

File handling is used to store data permanently in files. Common modes are `r` for reading, `w` for writing, and `a` for appending.

## Common File Modes

| Mode | Meaning |
| --- | --- |
| `r` | Read an existing file |
| `w` | Write to a file. Creates a new file or clears an existing file |
| `a` | Append data at the end of a file |
| `x` | Create a new file. Gives error if file already exists |

````python
# Example 1: Creating and writing in a text file
file = open("students.txt", "w")
file.write("Aman\n")
file.write("Riya\n")
file.write("Kabir\n")
file.close()

print("students.txt created successfully")
````

````python
# Example 2: Reading the full file using read()
file = open("students.txt", "r")
data = file.read()
file.close()

print(data)
````

````python
# Example 3: Reading a file line by line
file = open("students.txt", "r")

for line in file:
    print(line.strip())

file.close()
````

````python
# Example 4: Appending data to an existing file
file = open("students.txt", "a")
file.write("Meera\n")
file.write("Arjun\n")
file.close()

print("New names added successfully")
````

````python
# Example 5: Using with open()
# The file closes automatically after the block ends.
with open("students.txt", "r") as file:
    data = file.read()

print(data)
````

````python
# Example 6: Writing CSV-style marks data
with open("marks.csv", "w") as file:
    file.write("Roll No,Name,Marks\n")
    file.write("1,Aman,87\n")
    file.write("2,Riya,92\n")
    file.write("3,Kabir,79\n")

print("marks.csv created successfully")
````

````python
# Example 7: Reading CSV-style data line by line
with open("marks.csv", "r") as file:
    for line in file:
        roll_no, name, marks = line.strip().split(",")
        print(roll_no, name, marks)
````

## Exam Points

- Always close a file after using `open()`.
- `with open()` is safer because it closes the file automatically.
- `w` mode overwrites old data.
- `a` mode keeps old data and adds new data at the end.
- `read()` reads the whole file, while a `for` loop can read one line at a time.

# Database Management

These examples use `sqlite3`, which is available with Python. The same SQL ideas are used in MySQL also.

## Integrating SQL with Python

To connect SQL with Python, import a suitable module. Here we use `sqlite3`.

````python
import sqlite3

connection = sqlite3.connect("school.db")
cursor = connection.cursor()

def show_table(query):
    cursor.execute(query)
    for row in cursor.fetchall():
        print(row)
````

## Create a Student Table and Insert Data

````python
cursor.execute("DROP TABLE IF EXISTS student")

cursor.execute("""
CREATE TABLE student (
    roll_no INTEGER PRIMARY KEY,
    name TEXT,
    class TEXT,
    marks INTEGER,
    city TEXT
)
""")

records = [
    (1, "Aman", "12A", 87, "Delhi"),
    (2, "Riya", "12A", 92, "Mumbai"),
    (3, "Kabir", "12B", 79, "Delhi"),
    (4, "Meera", "12B", 95, "Kolkata"),
    (5, "Arjun", "12A", 68, "Mumbai")
]

cursor.executemany("INSERT INTO student VALUES (?, ?, ?, ?, ?)", records)
connection.commit()

show_table("SELECT * FROM student")
````

## ALTER TABLE: Add Attribute, Modify Data Type, Drop Attribute

Common school SQL syntax:

```sql
ALTER TABLE student ADD email VARCHAR(50);
ALTER TABLE student MODIFY marks FLOAT;
ALTER TABLE student DROP COLUMN email;
```

SQLite supports `ADD COLUMN`, but changing data type is usually done by creating a new table.

````python
# Add a new attribute
cursor.execute("ALTER TABLE student ADD COLUMN email TEXT")

cursor.execute("UPDATE student SET email = 'aman@example.com' WHERE roll_no = 1")
cursor.execute("UPDATE student SET email = 'riya@example.com' WHERE roll_no = 2")
connection.commit()

show_table("SELECT * FROM student")
````

````python
# Modify marks data type and drop email by recreating the table in SQLite
cursor.execute("""
CREATE TABLE student_new (
    roll_no INTEGER PRIMARY KEY,
    name TEXT,
    class TEXT,
    marks REAL,
    city TEXT
)
""")

cursor.execute("""
INSERT INTO student_new (roll_no, name, class, marks, city)
SELECT roll_no, name, class, CAST(marks AS REAL), city FROM student
""")

cursor.execute("DROP TABLE student")
cursor.execute("ALTER TABLE student_new RENAME TO student")
connection.commit()

show_table("PRAGMA table_info(student)")
````

## UPDATE, ORDER BY, and DELETE

````python
# UPDATE table to modify data
cursor.execute("UPDATE student SET marks = 90 WHERE roll_no = 3")
connection.commit()

show_table("SELECT * FROM student")
````

````python
# ORDER BY in ascending order
print("Marks in ascending order:")
show_table("SELECT * FROM student ORDER BY marks ASC")

print("\nMarks in descending order:")
show_table("SELECT * FROM student ORDER BY marks DESC")
````

````python
# DELETE tuple where marks are less than 70
cursor.execute("DELETE FROM student WHERE marks < 70")
connection.commit()

show_table("SELECT * FROM student")
````

## GROUP BY with MIN, MAX, SUM, COUNT, and AVG

````python
show_table("""
SELECT city,
       MIN(marks),
       MAX(marks),
       SUM(marks),
       COUNT(*),
       AVG(marks)
FROM student
GROUP BY city
""")
````

## Similar Exercise

Create an `employee` table with employee id, name, department, salary, and city. Practice `ALTER`, `UPDATE`, `ORDER BY`, `DELETE`, and `GROUP BY` commands on it.

````python
cursor.execute("DROP TABLE IF EXISTS employee")

cursor.execute("""
CREATE TABLE employee (
    emp_id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    salary INTEGER,
    city TEXT
)
""")

employees = [
    (101, "Neha", "Sales", 35000, "Delhi"),
    (102, "Rahul", "IT", 50000, "Mumbai"),
    (103, "Sara", "IT", 55000, "Delhi"),
    (104, "Vikram", "Sales", 40000, "Kolkata")
]

cursor.executemany("INSERT INTO employee VALUES (?, ?, ?, ?, ?)", employees)
connection.commit()

print("Employees ordered by salary:")
show_table("SELECT * FROM employee ORDER BY salary DESC")

print("\nDepartment wise salary details:")
show_table("""
SELECT department, MIN(salary), MAX(salary), SUM(salary), COUNT(*), AVG(salary)
FROM employee
GROUP BY department
""")
````
