```python
#Program-1
# There are 25 students in a class. Their heights are given below. Write a Python Program to
# find the mean.
#heights → 145, 151, 152, 149, 147, 152, 151,149, 152, 151, 147, 148, 155, 147,152,151,
#149,145, 147, 152,146, 148, 150, 152, 151

import statistics
heights = [145, 151, 152, 149, 147, 152, 151, 149, 152, 151, 147, 148, 155, 147, 152, 151,
           149, 145, 147, 152, 146, 148, 150, 152, 151] 

print("Mean height of the students is:", statistics.mean(heights))
```

```python
# Program -6
# Write a program to draw a line chart to visualize the comparative rainfall data for Nanded district.
#  data has been downloaded from data.gov.in


import pandas as pd
import matplotlib.pyplot as plt 
df=pd.read_csv("dailyRainfallReportNanded.csv") 
df
```

```python
x=df['Date' ] 
y=df['Daily Rainfall (m.m)']

# plt.figure(figsize=(6,4))
plt.plot(x,y,marker='o', markersize=8, markeredgecolor='red', markerfacecolor='yellow' )
plt.xticks(rotation = 45)

plt.xlabel( "Date", fontname='Calibri' ,color='b' ,fontsize=12)
plt.ylabel("Rainfall(mm)",fontname='Calibri' ,color='b' ,fontsize=12)
plt.title("Rainfall data of Nanded District", fontname='Calibri' ,color='m' ,fontsize=16)
plt.show()
```

```python
# Program 7
# Write a program to draw a bar chart to visualize the
# comparative rainfall data for 12 months in Tamil Nadu
# using the CSV file "rainfall.csv".

import pandas as pd
import matplotlib.pyplot as plt 
df=pd.read_csv("rainfall.csv") 
x=df['Months']
y=df[ 'Rainfall(cm)']
plt.c=["r","g","b","y","m","c"]
plt.figure(figsize=(6,4))
plt.bar(x,y,color=c,edgecolor='k')
plt.xticks(rotation = 45)
plt.xlabel("Months", fontname='Calibri',color='b' ,fontsize=12)
plt.ylabel("Rainfall(cm)",fontname='Calibri',color='b' ,fontsize=12)
plt.title("Rainfall data of Tamil Nadu", fontname='Calibri',color='m' ,fontsize=16)
plt.show()

```
