# Chapter 5: Control Flow

    **Level:** Beginner to Intermediate  
    **Understand before moving on:** if, elif, else, for, while, range, enumerate, zip, break, continue, pass, loop else, match and case

    ## Core Coverage

    - `if`, `elif`, `else`
- nested conditions
- `for` loops

- `range`
- `enumerate`
- `zip`
- `break` and `continue`
- `pass`
- loop `else`
- `match` and `case`

    > **What to remember**  
    > Control flow decides which code runs and how often.


````python
distance = 125
if distance <= 50:
    decision = 'HOLD'
elif distance <= 200:
    decision = 'RECENTER'
else:
    decision = 'EMERGENCY'
print(decision)

strikes = [22000, 22050, 22100]
for position, strike in enumerate(strikes):
    print(position, strike)

for symbol, step in zip(['NIFTY', 'BANKNIFTY'], [50, 100]):
    print(symbol, step)

count_value = 0
while count_value < 5:
    count_value += 1
    if count_value == 2:
        continue
    if count_value == 4:
        break
    print('processed', count_value)
else:
    print('loop ended without break')

message_type = 'tick'
match message_type:
    case 'tick':
        action = 'process tick'
    case 'heartbeat':
        action = 'keep alive'
    case 'error':
        action = 'log and reconnect'
    case _:
        action = 'unknown'
print(action)
````

> **Common mistake**  
> `case variable_name:` captures any value into a new name. It does not compare with an existing variable unless you use more advanced pattern matching carefully.


````python
for strike in [22000, 22050, 22100, 22150]:
    if strike < 22100:
        continue
    print(strike)
````

````python
def calculate_atm(price, step):
    """Return the nearest strike for a price and strike step."""
    rounded_units = round(price / step)
    return rounded_units * step

print(calculate_atm(22137.5, 50))
print(type(calculate_atm))


def build_contract_key(symbol, expiry, strike, option_type='CE'):
    """Build a readable educational option key."""
    return f'{symbol}:{expiry}:{strike}:{option_type}'

print(build_contract_key('NIFTY', '2026-06-25', 22200))
print(build_contract_key('NIFTY', '2026-06-25', 22200, option_type='PE'))


def strike_band(atm, step, count_each_side):
    """Return lower and upper strike boundaries."""
    return atm - step * count_each_side, atm + step * count_each_side

lower, upper = strike_band(22100, 50, 2)
print(lower, upper, type(strike_band(22100, 50, 2)))
````

## Signature Breakdown

```python
async def subscribe_instruments(websocket, instrument_keys: list[str], mode: str | None = None):
```

- `async def` defines an asynchronous function; full behavior appears in Chapter 18.
- `websocket` is a parameter.
- `instrument_keys: list[str]` is a type hint meaning list of strings.
- `mode: str | None = None` means the argument may be a string or `None`, and defaults to `None`.
- Parentheses group parameters; they do not create a tuple here.


````python
def normalize_strike(strike, /, *, step=50):
    """Normalize strike with positional-only strike and keyword-only step."""
    return round(strike / step) * step

print(normalize_strike(22137, step=50))


def log_event(event_name, *args, **kwargs):
    """Show flexible positional and keyword arguments."""
    print('event:', event_name)
    print('args:', args, type(args))
    print('kwargs:', kwargs, type(kwargs))

log_event('subscription', 'NIFTY', 'BANKNIFTY', mode='full', retry=1)


def make_threshold_checker(threshold):
    """Return a function that remembers threshold."""
    def is_far(distance):
        return distance > threshold
    return is_far

checker = make_threshold_checker(100)
print(checker(125))


def select_strikes(atm, step, count_each_side):
    """Return strikes centered around ATM."""
    strikes = []
    for offset in range(-count_each_side, count_each_side + 1):
        strikes.append(atm + offset * step)
    return strikes

print(select_strikes(22100, 50, 2))
````

## `LOG_DIR.mkdir(parents=True, exist_ok=True)` Word by Word

- `LOG_DIR` is a `Path` object.
- `.mkdir` is method lookup.
- `(...)` calls the method.
- `parents=True` creates missing parent directories.
- `exist_ok=True` avoids failure if the directory already exists.

Directory creation order matters because child directories need parent directories unless `parents=True` creates them.


````python
with tempfile.TemporaryDirectory() as tmpdir:
    base = Path(tmpdir)
    output_dir = base / 'market_course_output'
    log_dir = output_dir / 'logs'
    log_dir.mkdir(parents=True, exist_ok=True)
    print(output_dir.exists(), log_dir.exists())

    settings_path = output_dir / 'settings.txt'
    settings_path.write_text('symbol=NIFTY\ntimezone=Asia/Kolkata\n', encoding='utf-8')
    print(settings_path.name, settings_path.parent, settings_path.suffix, settings_path.stem)
    print(settings_path.read_text(encoding='utf-8'))

    config_path = output_dir / 'config.json'
    config_data = {'symbol': 'NIFTY', 'mode': 'full', 'strike_step': 50}
    config_path.write_text(json.dumps(config_data, indent=2), encoding='utf-8')
    print(json.loads(config_path.read_text(encoding='utf-8')))

    csv_path = output_dir / 'instruments.csv'
    with csv_path.open('w', newline='', encoding='utf-8') as file:
        writer = csv.DictWriter(file, fieldnames=['key', 'strike'])
        writer.writeheader()
        writer.writerows([{'key': 'NIFTY-22100-CE', 'strike': '22100'}])
    with csv_path.open('r', newline='', encoding='utf-8') as file:
        print(list(csv.DictReader(file)))
````

````python
def write_text_atomically(path, text):
    """Write text through a temporary sibling file and atomic replacement."""
    temporary_path = path.with_suffix(path.suffix + '.tmp')
    temporary_path.write_text(text, encoding='utf-8')
    os.replace(temporary_path, path)

with tempfile.TemporaryDirectory() as tmpdir:
    out_path = Path(tmpdir) / 'latest_tick.json'
    write_text_atomically(out_path, json.dumps({'symbol': 'NIFTY', 'price': 22100.5}))
    print(out_path.read_text(encoding='utf-8'))
````

````python
def greet(name):
    """Return a greeting."""
    return f'Hello {name}'

another_name = greet
print(another_name('NIFTY'))


def make_prefixer(prefix):
    """Return a function that remembers prefix."""
    def prefixer(text):
        return f'{prefix}:{text}'
    return prefixer

print(make_prefixer('MARKET')('tick'))


def log_calls(function):
    """Decorator preserving metadata."""
    @wraps(function)
    def wrapper(*args, **kwargs):
        print(f'calling {function.__name__}')
        return function(*args, **kwargs)
    return wrapper

@log_calls
def add_prices(first, second):
    """Add two prices."""
    return first + second

print(add_prices(100.5, 2.5))
print(add_prices.__name__)
````

`@log_calls` is approximately equivalent to `add_prices = log_calls(add_prices)`. `wraps` keeps the original function name and docstring.


````python
def retry(times):
    """Return a decorator that retries ValueError failures."""
    def decorator(function):
        @wraps(function)
        def wrapper(*args, **kwargs):
            last_error = None
            for attempt in range(1, times + 1):
                try:
                    return function(*args, **kwargs)
                except ValueError as error:
                    last_error = error
                    print('attempt failed', attempt)
            raise last_error
        return wrapper
    return decorator

attempt_counter = {'count': 0}
@retry(times=3)
def flaky_operation():
    """Fail once, then succeed."""
    attempt_counter['count'] += 1
    if attempt_counter['count'] < 2:
        raise ValueError('temporary failure')
    return 'success'

print(flaky_operation())
````

## Required Logging Block, Line by Line

`LOG_DIR = OUTPUT_DIR / 'logs'` assigns a `Path` made by path joining. `LOG_DIR.mkdir(parents=True, exist_ok=True)` creates the directory and missing parents. `logging.getLogger('market_stream')` is a logging API call. `logger.setLevel(logging.INFO)` sets minimum severity. `logger.propagate = False` prevents records from bubbling to parent loggers. `logging.Formatter(...)`, `StreamHandler`, and `RotatingFileHandler` create logging API objects. `logger.addHandler(...)` attaches a destination.

Rerunning a setup cell can add duplicate handlers, so the safe version removes old handlers first.


````python
def build_notebook_safe_logger(output_dir):
    """Create a logger safely even when rerun in a notebook."""
    log_dir = output_dir / 'logs'
    output_dir.mkdir(parents=True, exist_ok=True)
    log_dir.mkdir(parents=True, exist_ok=True)

    logger = logging.getLogger('market_stream_notebook_safe')
    logger.setLevel(logging.INFO)
    logger.propagate = False

    for handler in list(logger.handlers):
        logger.removeHandler(handler)
        handler.close()

    formatter = logging.Formatter(
        fmt='%(asctime)s | %(levelname)s | %(threadName)s | %(message)s',
        datefmt='%Y-%m-%d %H:%M:%S',
    )
    console = logging.StreamHandler()
    console.setFormatter(formatter)
    logger.addHandler(console)

    file_handler = RotatingFileHandler(log_dir / 'stream.log', maxBytes=10 * 1024 * 1024, backupCount=5, encoding='utf-8')
    file_handler.setFormatter(formatter)
    logger.addHandler(file_handler)
    return logger

with tempfile.TemporaryDirectory() as tmpdir:
    logger = build_notebook_safe_logger(Path(tmpdir) / 'output')
    logger.info('logger ready')
    print('handler count:', len(logger.handlers))

try:
    parse_strike('bad-strike')
except ValueError:
    logger.exception('Failed to parse strike')
````

````python
valid_keys_list = [f'KEY-{number}' for number in range(10_000)]
valid_keys_set = set(valid_keys_list)
needle = 'KEY-9999'
print(needle in valid_keys_list)
print(needle in valid_keys_set)
print('list:', round(timeit.timeit(lambda: needle in valid_keys_list, number=1000), 6))
print('set:', round(timeit.timeit(lambda: needle in valid_keys_set, number=1000), 6))


def batched(items, batch_size):
    """Yield batches of at most batch_size."""
    batch = []
    for item in items:
        batch.append(item)
        if len(batch) == batch_size:
            yield batch
            batch = []
    if batch:
        yield batch

print(list(batched(range(7), 3)))

profiler = cProfile.Profile()
profiler.enable()
select_strikes(22100, 50, 10)
profiler.disable()
stream = io.StringIO()
pstats.Stats(profiler, stream=stream).sort_stats('cumtime').print_stats(3)
print(stream.getvalue())
````

````python
def calculate(first: float, operator_text: str, second: float) -> float:
    """Calculate a basic arithmetic result."""
    if operator_text == '+':
        return first + second
    if operator_text == '-':
        return first - second
    if operator_text == '*':
        return first * second
    if operator_text == '/':
        if second == 0:
            raise ValueError('Cannot divide by zero')
        return first / second
    raise ValueError(f'Unsupported operator: {operator_text}')

def calculator_main(argv: list[str]) -> float:
    """Command-line-style entry point."""
    if len(argv) != 3:
        raise ValueError('Usage: NUMBER OPERATOR NUMBER')
    return calculate(float(argv[0]), argv[1], float(argv[2]))

assert calculator_main(['10', '+', '5']) == 15
assert calculator_main(['10', '*', '5']) == 50
print('calculator:', calculator_main(['22', '/', '2']))


def load_stream_config(path: Path) -> dict[str, object]:
    """Load and validate stream config JSON."""
    config = json.loads(path.read_text(encoding='utf-8'))
    required = {'symbol', 'mode', 'strike_step'}
    missing = required - set(config)
    if missing:
        raise ConfigurationError(f'Missing config keys: {sorted(missing)}')
    return config

with tempfile.TemporaryDirectory() as tmpdir:
    config_path = Path(tmpdir) / 'config.json'
    config_path.write_text(json.dumps({'symbol': 'NIFTY', 'mode': 'full', 'strike_step': 50}), encoding='utf-8')
    loaded = load_stream_config(config_path)
    assert loaded['symbol'] == 'NIFTY'
    print('config:', loaded)
````

````python
def read_instruments_csv(path: Path) -> list[dict[str, str]]:
    """Read instruments from CSV."""
    with path.open('r', newline='', encoding='utf-8') as file:
        return list(csv.DictReader(file))


def select_instruments(instruments: list[dict[str, str]], symbol: str, min_strike: int, max_strike: int) -> list[dict[str, str]]:
    """Select instruments by symbol and strike range."""
    selected = []
    for instrument in instruments:
        strike = int(instrument['strike'])
        if instrument['symbol'] == symbol and min_strike <= strike <= max_strike:
            selected.append(instrument)
    return selected

with tempfile.TemporaryDirectory() as tmpdir:
    path = Path(tmpdir) / 'instruments.csv'
    rows = [
        {'symbol': 'NIFTY', 'strike': '22100', 'key': 'NIFTY-22100-CE'},
        {'symbol': 'NIFTY', 'strike': '22200', 'key': 'NIFTY-22200-CE'},
        {'symbol': 'BANKNIFTY', 'strike': '48000', 'key': 'BANKNIFTY-48000-CE'},
    ]
    with path.open('w', newline='', encoding='utf-8') as file:
        writer = csv.DictWriter(file, fieldnames=['symbol', 'strike', 'key'])
        writer.writeheader()
        writer.writerows(rows)
    selected = select_instruments(read_instruments_csv(path), 'NIFTY', 22000, 22150)
    assert len(selected) == 1
    print('selected:', selected)
````

````python
def process_ticks(ticks: list[dict[str, object]], logger: logging.Logger) -> list[dict[str, object]]:
    """Return ticks with positive numeric prices."""
    valid = []
    for tick in ticks:
        price = tick.get('price')
        if not isinstance(price, (int, float)) or price <= 0:
            logger.warning('Skipping invalid tick: %s', tick)
            continue
        valid.append(tick)
    logger.info('Processed %s valid ticks', len(valid))
    return valid

valid_ticks = process_ticks([{'price': 1.0}, {'price': -1.0}], logger)
assert valid_ticks == [{'price': 1.0}]


def option_strike_selector(price: float, step: int, count_each_side: int) -> list[int]:
    """Select strikes around nearest ATM."""
    atm = nearest_strike(price, step)
    return select_strikes(atm, step, count_each_side)

assert option_strike_selector(22137.5, 50, 1) == [22100, 22150, 22200]

@dataclass
class Decision:
    action: str
    reason: str
    distance: float


def decide_recenter(current_price: float, center_price: float, recenter_threshold: float, emergency_threshold: float) -> Decision:
    """Return recenter decision based on distance from center."""
    distance = abs(current_price - center_price)
    if distance >= emergency_threshold:
        return Decision('EMERGENCY', 'emergency threshold reached', distance)
    if distance >= recenter_threshold:
        return Decision('RECENTER', 'recenter threshold reached', distance)
    return Decision('HOLD', 'inside band', distance)

assert decide_recenter(22350, 22100, 100, 200).action == 'EMERGENCY'
print(valid_ticks)
print(option_strike_selector(22137.5, 50, 2))
print(decide_recenter(22225, 22100, 100, 200))
````

````python
def run_queue_demo() -> list[dict[str, int]]:
    """Run producer-consumer queue demo."""
    queue = Queue()
    stop = Event()
    consumed = []
    def produce():
        for sequence in range(3):
            queue.put({'sequence': sequence})
        stop.set()
    def consume():
        while not stop.is_set() or not queue.empty():
            try:
                consumed.append(queue.get(timeout=0.1))
            except Empty:
                continue
    producer_thread = Thread(target=produce)
    consumer_thread = Thread(target=consume)
    producer_thread.start()
    consumer_thread.start()
    producer_thread.join()
    consumer_thread.join()
    return consumed

queue_result = run_queue_demo()
assert len(queue_result) == 3
print(queue_result)

async def mock_market_stream(symbol: str, tick_count: int):
    """Yield mock market ticks asynchronously."""
    for sequence in range(tick_count):
        await asyncio.sleep(0.01)
        yield {'symbol': symbol, 'price': 22100 + sequence, 'sequence': sequence}

async def collect_stream(symbol: str, tick_count: int) -> list[dict[str, object]]:
    """Collect async stream for testing."""
    collected = []
    async for tick in mock_market_stream(symbol, tick_count):
        collected.append(tick)
    return collected

stream_ticks = await collect_stream('NIFTY', 3)
assert len(stream_ticks) == 3
print(stream_ticks)
````

````python
class TickRecorder:
    """Record ticks in DuckDB when available, otherwise memory."""
    def __init__(self):
        self.rows = []
        self.connection = None
        if HAS_DUCKDB:
            self.connection = duckdb.connect(database=':memory:')
            self.connection.execute('CREATE TABLE ticks(symbol VARCHAR, price DOUBLE, sequence INTEGER)')
    def add_tick(self, tick: dict[str, object]) -> None:
        if self.connection is not None:
            self.connection.execute('INSERT INTO ticks VALUES (?, ?, ?)', [tick['symbol'], tick['price'], tick['sequence']])
        else:
            self.rows.append(dict(tick))
    def count(self) -> int:
        if self.connection is not None:
            return self.connection.execute('SELECT COUNT(*) FROM ticks').fetchone()[0]
        return len(self.rows)
    def close(self) -> None:
        if self.connection is not None:
            self.connection.close()

recorder = TickRecorder()
recorder.add_tick({'symbol': 'NIFTY', 'price': 22100.5, 'sequence': 1})
assert recorder.count() == 1
print('recorded ticks:', recorder.count())
recorder.close()

class TestedSubscriptionManager:
    """Manage subscription keys and payloads."""
    def __init__(self):
        self._keys: set[str] = set()
    def add(self, key: str) -> None:
        if not key:
            raise ValueError('key must not be empty')
        self._keys.add(key)
    def remove(self, key: str) -> None:
        self._keys.discard(key)
    def keys(self) -> list[str]:
        return sorted(self._keys)
    def payload(self, mode: str = 'full') -> str:
        return json.dumps({'mode': mode, 'keys': self.keys()})

sub_manager = TestedSubscriptionManager()
sub_manager.add('NIFTY')
sub_manager.add('NIFTY')
sub_manager.add('BANKNIFTY')
assert sub_manager.keys() == ['BANKNIFTY', 'NIFTY']
print(sub_manager.payload())
````

````python
@dataclass(frozen=True)
class AppConfig:
    """Configuration for the educational market stream."""
    symbol: str = 'NIFTY'
    strike_step: int = 50
    strikes_each_side: int = 1
    mode: str = 'full'
    recenter_threshold: float = 75.0
    emergency_threshold: float = 150.0
    max_messages: int = 5

@dataclass(frozen=True)
class AppInstrument:
    """Simplified instrument record."""
    key: str
    symbol: str
    strike: int
    side: str

@dataclass(frozen=True)
class AppTick:
    """Validated market tick."""
    symbol: str
    price: float
    sequence: int


def app_build_logger() -> logging.Logger:
    """Build a console-only notebook-safe logger."""
    app_logger = logging.getLogger('educational_market_stream')
    app_logger.setLevel(logging.INFO)
    app_logger.propagate = False
    for handler in list(app_logger.handlers):
        app_logger.removeHandler(handler)
        handler.close()
    handler = logging.StreamHandler()
    handler.setFormatter(logging.Formatter('%(levelname)s | %(message)s'))
    app_logger.addHandler(handler)
    return app_logger


def app_load_instruments(config: AppConfig) -> list[AppInstrument]:
    """Generate mock instruments around a center strike."""
    center = 22100
    instruments = []
    for strike in select_strikes(center, config.strike_step, config.strikes_each_side + 1):
        for side in ['CE', 'PE']:
            instruments.append(AppInstrument(f'{config.symbol}-{strike}-{side}', config.symbol, strike, side))
    return instruments


def app_select_subscription_keys(instruments: list[AppInstrument], center_strike: int, config: AppConfig) -> list[str]:
    """Select subscription keys around center strike."""
    min_strike = center_strike - config.strike_step * config.strikes_each_side
    max_strike = center_strike + config.strike_step * config.strikes_each_side
    return sorted(instrument.key for instrument in instruments if min_strike <= instrument.strike <= max_strike)


def app_validate_message(message: dict[str, object]) -> AppTick:
    """Validate a raw message and return an AppTick."""
    try:
        symbol = str(message['symbol'])
        price = float(message['price'])
        sequence = int(message['sequence'])
    except KeyError as error:
        raise ValueError(f'missing message field: {error.args[0]}') from error
    except (TypeError, ValueError) as error:
        raise ValueError(f'invalid message: {message!r}') from error
    if price <= 0:
        raise ValueError(f'price must be positive: {price}')
    return AppTick(symbol, price, sequence)


def app_decide_recenter(tick: AppTick, center_price: float, config: AppConfig) -> Decision:
    """Return recenter decision for a tick."""
    return decide_recenter(tick.price, center_price, config.recenter_threshold, config.emergency_threshold)

class AppTickStore:
    """Store ticks in DuckDB when available, otherwise memory."""
    def __init__(self):
        self.connection = None
        self.rows: list[AppTick] = []
        if HAS_DUCKDB:
            self.connection = duckdb.connect(database=':memory:')
            self.connection.execute('CREATE TABLE ticks(symbol VARCHAR, price DOUBLE, sequence INTEGER)')
    def add(self, tick: AppTick) -> None:
        if self.connection is None:
            self.rows.append(tick)
        else:
            self.connection.execute('INSERT INTO ticks VALUES (?, ?, ?)', [tick.symbol, tick.price, tick.sequence])
    def count(self) -> int:
        if self.connection is None:
            return len(self.rows)
        return self.connection.execute('SELECT COUNT(*) FROM ticks').fetchone()[0]
    def close(self) -> None:
        if self.connection is not None:
            self.connection.close()

async def app_simulated_stream(config: AppConfig):
    """Yield generated market messages."""
    for sequence in range(config.max_messages):
        await asyncio.sleep(0.01)
        yield {'symbol': config.symbol, 'price': 22100.0 + sequence * 25.0, 'sequence': sequence}

async def app_run(config: AppConfig) -> dict[str, object]:
    """Run the educational market-stream app."""
    app_logger = app_build_logger()
    instruments = app_load_instruments(config)
    center_strike = 22100
    keys = app_select_subscription_keys(instruments, center_strike, config)
    app_logger.info('selected %s subscription keys', len(keys))

    queue = asyncio.Queue()
    store = AppTickStore()
    decisions: list[Decision] = []

    async def producer() -> None:
        async for raw_message in app_simulated_stream(config):
            try:
                tick = app_validate_message(raw_message)
            except ValueError as error:
                app_logger.warning('invalid message skipped: %s', error)
                continue
            await queue.put(tick)

    async def consumer() -> None:
        processed = 0
        while processed < config.max_messages:
            try:
                tick = await asyncio.wait_for(queue.get(), timeout=1.0)
            except asyncio.TimeoutError:
                app_logger.warning('consumer timed out')
                break
            store.add(tick)
            decisions.append(app_decide_recenter(tick, center_strike, config))
            queue.task_done()
            processed += 1

    try:
        await asyncio.gather(producer(), consumer())
        return {'keys': keys, 'stored_count': store.count(), 'decisions': decisions}
    finally:
        store.close()

final_result = await app_run(AppConfig(max_messages=5))
print(final_result)
assert len(final_result['keys']) == 6
assert final_result['stored_count'] == 5
assert any(decision.action in {'RECENTER', 'EMERGENCY'} for decision in final_result['decisions'])
print('final integrated project tests passed')
````
